```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("/content/sample_data/diabetes.csv")

# Display first five records
print("First Five Records:")
print(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
print(df.info())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Descriptive Statistics:")
print(df.describe())
```

```python
print("Mean of Numerical Columns:")
print(df.mean(numeric_only=True))
```

```python
print("Median of Numerical Columns:")
print(df.median(numeric_only=True))
```

```python
print("Mode of Numerical Columns:")
print(df.mode())
```

```python
print("Minimum Values:")
print(df.min(numeric_only=True))
```

```python
print("Maximum Values:")
print(df.max(numeric_only=True))
```

```python
print("Variance:")
print(df.var(numeric_only=True))
```

```python
print("Standard Deviation:")
print(df.std(numeric_only=True))
```

```python
df.hist(figsize=(12,10))

plt.suptitle("Histograms of Numerical Variables")
plt.show()
```

```python
plt.figure(figsize=(12,6))

df.boxplot()

plt.xticks(rotation=45)
plt.title("Boxplots of Numerical Variables")
plt.show()
```

```python
outcome_counts = df["Outcome"].value_counts()

plt.figure(figsize=(6,5))

outcome_counts.plot(kind="bar")

plt.title("Frequency of Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.scatter(df["Glucose"], df["BMI"])

plt.title("Glucose vs BMI")
plt.xlabel("Glucose")
plt.ylabel("BMI")

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.scatter(df["Age"], df["Glucose"])

plt.title("Age vs Glucose")
plt.xlabel("Age")
plt.ylabel("Glucose")

plt.show()
```

```python
import seaborn as sns

selected_columns = ["Glucose", "BMI", "Age", "Insulin", "Outcome"]

sns.pairplot(df[selected_columns], hue="Outcome")

plt.show()
```
